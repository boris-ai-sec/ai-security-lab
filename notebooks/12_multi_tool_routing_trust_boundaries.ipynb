# Lab12 — Multi-Tool Routing & Trust Boundaries

This lab explores routing logic, trust boundaries, workflow branching, and authority propagation in AI-agent systems.

Main focus:

- multi-tool orchestration
- routing visibility
- trust boundaries
- workflow branching
- execution propagation
- orchestration transparency
- authority separation

```python
tool_registry = {
    "search_tool": {
        "trust_boundary": "external_tool",
        "risk_level": "low"
    },

    "payment_tool": {
        "trust_boundary": "restricted_execution",
        "risk_level": "high"
    },

    "memory_tool": {
        "trust_boundary": "internal_memory",
        "risk_level": "medium"
    }
}
```

```python
for tool, metadata in tool_registry.items():
    print(tool, metadata)
```

```python
def route_request(tool_name):

    tool_data = tool_registry[tool_name]

    print("Routing request to:", tool_name)
    print("Trust boundary:", tool_data["trust_boundary"])
    print("Risk level:", tool_data["risk_level"])
```

```python
route_request("search_tool")
```

```python
def evaluate_routing(tool_name):

    tool_data = tool_registry[tool_name]

    if tool_data["risk_level"] == "high":
        print("Routing blocked: high-risk tool")

    else:
        print("Routing approved")
```

```python
evaluate_routing("search_tool")
```

```python
evaluate_routing("payment_tool")
```

```python
def inspect_boundary(tool_name):

    tool_data = tool_registry[tool_name]

    print("Tool:", tool_name)
    print("Boundary type:", tool_data["trust_boundary"])

    if tool_data["trust_boundary"] == "external_tool":
        print("External boundary crossing detected")

    elif tool_data["trust_boundary"] == "restricted_execution":
        print("Restricted execution boundary detected")

    else:
        print("Internal system boundary")
```

```python
inspect_boundary("search_tool")
```

```python
def secure_route(tool_name):

    tool_data = tool_registry[tool_name]

    print("Routing tool:", tool_name)
    print("Boundary:", tool_data["trust_boundary"])
    print("Risk:", tool_data["risk_level"])

    if tool_data["risk_level"] == "high":
        print("Routing denied")

    else:
        print("Routing approved")
```

```python
secure_route("memory_tool")
```

```python
def explainable_route(tool_name):

    tool_data = tool_registry[tool_name]

    print("Tool:", tool_name)
    print("Boundary:", tool_data["trust_boundary"])
    print("Risk:", tool_data["risk_level"])

    if tool_data["risk_level"] == "high":
        print("Routing denied")
        print("Reason: high-risk execution requires restriction")

    else:
        print("Routing approved")
        print("Reason: acceptable routing risk")
```

```python
explainable_route("payment_tool")
```

```python
workflow_chain = [
    "search_tool",
    "memory_tool",
    "payment_tool"
]
```

```python
for tool in workflow_chain:
    print("Workflow step:", tool)
```

```python
for tool in workflow_chain:

    tool_data = tool_registry[tool]

    print("Tool:", tool)
    print("Risk:", tool_data["risk_level"])

    if tool_data["risk_level"] == "high":
        print("High-risk propagation detected")

    print("---")
```

```python
for tool in workflow_chain:

    tool_data = tool_registry[tool]

    print("Workflow tool:", tool)
    print("Risk:", tool_data["risk_level"])

    if tool_data["risk_level"] == "high":
        print("Propagation stopped")
        break

    print("Propagation allowed")
    print("---")
```

```python
for tool in workflow_chain:

    tool_data = tool_registry[tool]

    print("Workflow tool:", tool)

    if tool_data["risk_level"] == "high":
        print("Workflow interrupted at:", tool)
        break

    print("Workflow continued")
```

```python
previous_boundary = None

for tool in workflow_chain:

    tool_data = tool_registry[tool]
    current_boundary = tool_data["trust_boundary"]

    print("Tool:", tool)
    print("Boundary:", current_boundary)

    if previous_boundary is not None:
        print("Boundary transition:",
              previous_boundary,
              "→",
              current_boundary)

    previous_boundary = current_boundary

    print("---")
```

```python
risk_order = {
    "low": 1,
    "medium": 2,
    "high": 3
}

previous_risk = None

for tool in workflow_chain:

    tool_data = tool_registry[tool]
    current_risk = tool_data["risk_level"]

    print("Tool:", tool)
    print("Risk:", current_risk)

    if previous_risk is not None:
        if risk_order[current_risk] > risk_order[previous_risk]:
            print("Authority escalation detected")

    previous_risk = current_risk

    print("---")
```

```python
highest_risk = "low"
workflow_status = "completed"
interrupted_at = None

for tool in workflow_chain:

    tool_data = tool_registry[tool]
    current_risk = tool_data["risk_level"]

    if risk_order[current_risk] > risk_order[highest_risk]:
        highest_risk = current_risk

    if current_risk == "high":
        workflow_status = "interrupted"
        interrupted_at = tool
        break

print("Workflow summary")
print("Highest risk reached:", highest_risk)
print("Workflow status:", workflow_status)

if interrupted_at is not None:
    print("Interrupted at:", interrupted_at)
```
