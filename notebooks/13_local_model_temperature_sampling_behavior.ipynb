## Observations

### Temperature 0.1

- responses were compact and highly deterministic
- wording remained conservative and predictable
- semantic drift was minimal
- output structure was tightly bounded

### Temperature 1.2

- responses became more exploratory
- semantic expansion increased
- sentence structure became less constrained
- continuation tendency became more noticeable

### Temperature 1.8

- abstraction and semantic looseness increased further
- outputs became less tightly bounded
- wording variability increased
- exploratory continuation behavior became more visible

## Governance Insights

- temperature affects probabilistic exploration behavior
- higher temperature reduces output predictability and reproducibility
- bounded prompts and output limits help stabilize inference behavior
- hallucinations are not the only indicator of stochastic drift
- semantic looseness and abstraction drift appear earlier than catastrophic failure

## Conclusion

Temperature is not simply a “creativity slider.”

It directly affects:
- operational predictability
- response boundedness
- semantic stability
- reproducibility
- governance reliability
