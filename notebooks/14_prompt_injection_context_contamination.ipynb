# Lab14 — Prompt Injection & Context Contamination

## Goal

Study how prompt injection and hidden instructions can contaminate workflow behavior across AI orchestration layers.

Focus areas:

- direct prompt injection
- indirect prompt injection
- hidden instruction propagation
- retrieval contamination
- authority confusion
- trust-boundary violations
- workflow observability
- contamination reconstruction

## Core Research Question

How do hidden instructions and contaminated context propagate through AI workflows, and how can governance-oriented observability improve detection and reconstruction?
