# Lab11 — AI Agent Observability & Decision Logging

Goal: Understand how observability, execution tracing, and governance logging improve visibility into AI-agent orchestration behavior.

Main focus:

- execution tracing
- governance visibility
- decision logging
- auditability
- orchestration transparency
- operational observability

```python
from datetime import datetime

execution_trace = []
```

```python
def log_execution(tool_name, action):

    event = {
        "timestamp": str(datetime.now()),
        "tool": tool_name,
        "action": action
    }

    execution_trace.append(event)
```

```python
log_execution("search_tool", "executed")

log_execution("payment_tool", "human_review")

log_execution("email_tool", "blocked")
```

```python
for event in execution_trace:
    print(event)
```

```python
def log_decision(tool_name, risk_level, decision):

    event = {
        "timestamp": str(datetime.now()),
        "tool": tool_name,
        "risk_level": risk_level,
        "decision": decision
    }

    execution_trace.append(event)
```

```python

for event in execution_trace[-3:]:
    print(event)
```

```python
log_decision("search_tool", "low", "approved")

log_decision("payment_tool", "medium", "human_review")

log_decision("email_tool", "high", "blocked")
```

```python
for event in execution_trace[-3:]:
    print(event)
```

```python
def log_governance_event(tool_name,
                         risk_level,
                         decision,
                         execution_status):

    event = {
        "timestamp": str(datetime.now()),
        "tool": tool_name,
        "risk_level": risk_level,
        "decision": decision,
        "execution_status": execution_status
    }

    execution_trace.append(event)
```

```python
log_governance_event(
    "search_tool",
    "low",
    "approved",
    "executed"
)

log_governance_event(
    "payment_tool",
    "medium",
    "human_review",
    "paused_for_human"
)

log_governance_event(
    "email_tool",
    "high",
    "blocked",
    "skipped"
)
```

```python
for event in execution_trace[-3:]:
    print(event)
```
