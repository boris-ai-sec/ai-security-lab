# Lab15 — Workflow Telemetry with Phoenix, OpenTelemetry, and Ollama

## Lab Objective

This lab explores how runtime telemetry can provide evidence for AI workflow reconstruction and AI governance review.

The lab starts with simple OpenTelemetry traces, then improves trace quality, and finally captures a real local model call through Ollama and displays it in Phoenix.

## Core Question

What can an AI Risk Review Framework conclude from runtime traces, and what remains unsupported by the available evidence?

## Framework Focus

This laboratory evaluates runtime telemetry as assessment evidence for AI Risk Review Framework, with emphasis on:

- Workflow reconstruction
- Evidence quality
- Evidence limitations
- Runtime observability
- Independent assessment capability

## Experiment 1 — Phoenix Installation and Smoke Test

### Goal

Verify that Phoenix can receive and visualize OpenTelemetry traces generated from Python.

### What was tested

- Phoenix local installation
- Phoenix local web UI
- OpenTelemetry trace export
- Basic span visibility
- Span attributes visibility

### Result

Phoenix successfully received and displayed telemetry traces from Python.

This confirms that the basic telemetry pipeline works:

Python workflow → OpenTelemetry → Phoenix server → Phoenix UI


### Evidence Produced

- Phoenix server successfully started.
- OpenTelemetry traces were received.
- Root spans were displayed.
- Child spans were displayed.
- Span attributes were accessible.
- Runtime workflow reconstruction became possible.

### Evidence Quality

Medium

Reason:
The telemetry proves workflow execution and span structure, but does not yet represent a real production AI workflow.

### Evidence Limitations

- Simulated workflow.
- No real LLM call.
- No retrieval.
- No tool execution.
- No production environment.

### Framework Interpretation

The Framework can conclude that runtime workflow reconstruction is technically possible.

The Framework cannot conclude anything about model quality, operational readiness, or production reliability from this evidence alone.

## Experiment 2 — Assessment-Friendly Trace (Trace v2)

### Goal

Improve the telemetry trace so that it becomes more useful for AI Risk Review rather than simple debugging.

### Changes Compared to Trace v1

- Added workflow metadata.
- Added input and output.
- Added evidence quality.
- Added evidence limitations.
- Added Framework interpretation fields.
- Added explicit status.

### Result

Trace v2 provides substantially richer assessment evidence than Trace v1.

Phoenix displays:

- workflow status;
- input;
- output;
- evidence metadata;
- assessment-oriented attributes.

# Laboratory Summary

## Key Findings

1. Phoenix was successfully installed and configured.
2. OpenTelemetry traces were exported successfully.
3. Runtime traces can be reconstructed visually.
4. Assessment-oriented telemetry provides more useful evidence than technical telemetry alone.
5. Phoenix successfully captured a real local Ollama model execution.

# Framework Notes

## What Framework Can Conclude

- Runtime workflow execution can be reconstructed.
- Workflow steps can be independently reviewed.
- Runtime evidence quality depends on telemetry design.
- Better span design improves assessment quality.

## What Framework Cannot Conclude

- Model correctness
- Model safety
- Production readiness
- RAG quality
- Agent reliability

Additional evidence is required before making these conclusions.
